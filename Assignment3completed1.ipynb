##Que. 1 
Create a Simple Linear Regression model for a given dataset. Also plot regression line.

Datasets:
1. NewspaperData.csv
Independent = ‘Daily’, Dependent=’Sunday’


```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
```

```python
df1=pd.read_csv(r"C:\Users\SUMEDH\Desktop\New folder\IIT-08-A-DS\Datasets\NewspaperData.csv")
```

```python
df1
```

```python

x = df1.drop(['Newspaper', 'sunday'], axis=1) 

y = df1['sunday']
```

```python
df1.head()
```

```python
x.head()
```

```python
y.head()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=12345)
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
model.predict([[391.952]])
```

```python
plt.scatter(df1['daily'],df1['sunday'],label ='sunday')
plt.plot(df1['daily'],model.predict(x),color="red", label ="best fit regression line")
plt.scatter(df1['daily'],model.predict(x),color ="red")
plt.xlabel("daily")
plt.ylabel("sunday")
plt.title("daily")
plt.legend()
```

```python
y_pred = model.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error
# mean absolute error (MAE)
mae = mean_absolute_error(y_test, y_pred)
mae
```

```python
from sklearn.metrics import mean_squared_error
mse = mean_squared_error(y_test, y_pred)
mse
```

```python
from sklearn.metrics import root_mean_squared_error
rmse = root_mean_squared_error(y_test, y_pred)
rmse
```
