##7. Write a Pandas program to add, subtract, multiple and divide two Pandas Series. Sample Series: [2, 4, 6, 8, 10], [1, 3, 5, 7, 9]

```python
import pandas as pd
num8 = [2,4,6,8,10]
num9 = [1,3,5,7,9]
series1 = pd.Series(num8)
series2 = pd.Series(num9)

result6 = series1 + series2
print(result6)
result7 = series1 - series2
print(result7)
result8 = series1 * series2
print(result8)
result9 = series1 % series2
print(result9)
```

##11. Consider mtcars.csv file  and plot the following graphs.
i) Line plot
ii) Bar Graph
iii) Pie Chart
iv) Boxplot
v) Pairplot
vi) Scatter Plot
vii) Histogram

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df1=pd.read_csv("C:\\Users\\SUMEDH\Desktop\\New folder\\IIT-08-A-DS\\Datasets\\mtcars.csv")
```

#line plot

```python
print(df1)
```

```python
plt.figure()
plt.plot(df1.index,df1['mpg'])
plt.xlabel('index')
plt.ylabel('mpg')
plt.title('line plot')
plt.show()
```

```python
avg_mpg = df1.groupby('cyl')['mpg'].mean()

plt.figure()
avg_mpg.plot(kind='bar')
plt.xlabel("Cyl")
plt.ylabel("average mpg")
plt.title("Bar Graph")
plt.show()
```

```python
cyl=df1['cyl'].value_counts()
plt.figure()
plt.pie(cyl,labels=cyl.index,autopct='%1.1f%%')
plt.title('pie chart')
plt.show()
```

```python
plt.figure()
plt.boxplot(df1['mpg'])
plt.ylabel('mpg')
plt.title('box plot')
plt.show()
```

##pair plot

```python
sns.pairplot(df1[['mpg','cyl','disp']])
plt.show()
```

##scatter plot

```python
plt.figure()
plt.scatter(df1['hp'],df1['mpg'])
plt.xlabel('hp')
plt.ylabel('mpg')
plt.title('scatter')
plt.show()
```

##histogram

```python
plt.figure()
plt.hist(df1['mpg'],bins=10)
plt.xlabel('mpg')
plt.ylabel('frequency')
plt.title('histogram')
plt.show()
```
